```python
sueldo = lambda h, v: h * v * 30

horas = float(input("Horas trabajadas por día: "))
valor = float(input("Valor por hora: "))

bruto = sueldo(horas, valor)

isss = bruto * 0.03
afp = bruto * 0.0725
isr = bruto * 0.10

neto = bruto - (isss + afp + isr)

print("Sueldo bruto:", bruto)
print("ISSS:", isss)
print("AFP:", afp)
print("ISR:", isr)
print("Sueldo neto:", neto)
```

```python
#FILTER
#dada una serie numero dame solo lso que son pares
serie = [2,5,6,8,9,10,12]
pares = filter(lambda num: num%2==0, serie)
impares = filter(lambda num: num%2!=0, serie)
print("Original: ", serie)
print("Pares:", list(pares))
print("Impares: ", list(impares))
```

```python
# Clasificar notas en aprobados y reprobados

aprobados = []
reprobados = []

n = int(input("¿Cuántas notas? "))

for i in range(n):
    nota = float(input("Nota: "))

    if nota >= 6:
        aprobados.append(nota)
    else:
        reprobados.append(nota)

print("Aprobados:", aprobados)
print("Reprobados:", reprobados)
```

```python
#ordenamiento.
lista = [8,1,9,5,3,2,4]
print("Original: ", lista)

lista.sort()
print("Ordenada:", lista)

#reversa
lista.reverse()
print("Reversa: ", lista)

valores = [8,7,1,9,4,5]
valores.sort(reverse=True)
print("Ordenada de mayor a menor: ", valores)
```
