## **4.	Sistema Academico usando JSON/diccionarios (20%)**

```python
import json
import math
from functools import reduce

class SistemaEstadistico:
    def __init__(self, datos_estudiantes):
        self.dataset = datos_estudiantes

    def obtener_columna(self, llave_campo):
        return [float(estudiante[llave_campo]) for estudiante in self.dataset.values()]

    def media_aritmetica(self, serie):
        if not serie: return 0
        suma = reduce(lambda acum, num: acum + num, serie, 0)
        return suma / len(serie)

    def mediana(self, serie):
        if not serie: return 0
        datos_ordenados = sorted(serie)
        n = len(datos_ordenados)
        mitad = n // 2

        if n % 2 == 0:
            return (datos_ordenados[mitad - 1] + datos_ordenados[mitad]) / 2
        else:
            return datos_ordenados[mitad]

    def frecuencia(self, serie):
        fre = {}
        for num in serie:
            fre[num] = fre.get(num, 0) + 1
        return fre

    def moda(self, serie):
        if not serie: return "No hay datos"
        frecuencia_datos = self.frecuencia(serie)
        max_frecuencia = max(frecuencia_datos.values())

        if max_frecuencia == 1:
            return "No hay moda (Amodal)"

        modas = [num for num, freq in frecuencia_datos.items() if freq == max_frecuencia]
        return modas

    def varianza(self, serie):
        if len(serie) < 2: return 0
        media = self.media_aritmetica(serie)
        suma_cuadrados = sum((x - media) ** 2 for x in serie)
        return suma_cuadrados / len(serie)

    def desviacion_estandar(self, serie):
        return math.sqrt(self.varianza(serie))

    def desviacion_tipica(self, serie):
        return self.desviacion_estandar(serie)

    def rango(self, serie):
        if not serie: return 0
        return max(serie) - min(serie)

    def calcular_todo(self, nombre_campo):
        serie = self.obtener_columna(nombre_campo)
        print(f"\n=== ANÁLISIS ESTADÍSTICO: {nombre_campo.upper()} ===")
        print(f"Media Aritmética: {self.media_aritmetica(serie):.2f}")
        print(f"Mediana:          {self.mediana(serie):.2f}")
        print(f"Moda:             {self.moda(serie)}")
        print(f"Varianza:         {self.varianza(serie):.2f}")
        print(f"Desviación Est.:  {self.desviacion_estandar(serie):.2f}")
        print(f"Desviación Típ.:  {self.desviacion_tipica(serie):.2f}")
        print(f"Rango:            {self.rango(serie):.2f}")
        print(f"Frecuencias:      {self.frecuencia(serie)}")


def explicar_diferencias():
    print("\nDIFERENCIA ENTRE LAS MEDIDAS ESTADÍSTICAS")
    print("""
Media aritmética: es el promedio de todos los datos, sumando
todo y dividiendo entre la cantidad de datos. Se ve muy
afectada por valores extremos.

Mediana: es el valor que queda en medio cuando los datos se
ordenan de menor a mayor. No le afectan los extremos, por
eso es más estable que la media.

Moda: es el valor que más se repite. Puede haber una, varias
o ninguna moda (amodal) si todos se repiten igual.

Varianza: mide qué tan dispersos están los datos respecto a
la media, pero queda en unidades al cuadrado.

Desviación estándar / desviación típica: es la raíz cuadrada
de la varianza. Son el mismo cálculo, solo dos nombres
distintos para lo mismo. Indica cuánto se aleja en promedio
cada dato del valor central.

Rango: es la diferencia entre el valor máximo y el mínimo.

Frecuencia: cuántas veces se repite cada valor en los datos.
""")


def guardar_dataset_json(dataset, ruta_archivo="dataset_ugb.json"):
    with open(ruta_archivo, "w", encoding="utf-8") as archivo:
        json.dump(dataset, archivo, ensure_ascii=False, indent=4)
    print(f"\nDataset guardado en '{ruta_archivo}'")


def cargar_dataset_json(ruta_archivo="dataset_ugb.json"):
    with open(ruta_archivo, "r", encoding="utf-8") as archivo:
        return json.load(archivo)


dataset_ugb = {
    "estudiante_1":  {"nombre": "Samuel Portillo",    "materia": "Programación_lll", "nota": 8.5, "edad": 20, "traslado_ugb": 45, "uso_telefono": 5},
    "estudiante_2":  {"nombre": "Danie  Quintanilla",       "materia": "Programación", "nota": 9.0, "edad": 19, "traslado_ugb": 20, "uso_telefono": 4},
    "estudiante_3":  {"nombre": "DAVID TURCIOS",      "materia": "Programación", "nota": 6.5, "edad": 20, "traslado_ugb": 60, "uso_telefono": 6},
    "estudiante_4":  {"nombre": "Mileidy González",     "materia": "Programación", "nota": 7.0, "edad": 19, "traslado_ugb": 15, "uso_telefono": 3},
    "estudiante_5":  {"nombre": "ALEXANDER JIMENEZ ",     "materia": "Programación", "nota": 10.0,"edad": 21, "traslado_ugb": 30, "uso_telefono": 5},
    "estudiante_6":  {"nombre": "ERNESTO RAMIREZ ",     "materia": "Programación", "nota": 5.5, "edad": 21, "traslado_ugb": 90, "uso_telefono": 7},
    "estudiante_7":  {"nombre": "Kevin Campos",     "materia": "Programación", "nota": 8.0, "edad": 19, "traslado_ugb": 40, "uso_telefono": 6},
    "estudiante_8":  {"nombre": "Julissa Parada",  "materia": "Programación", "nota": 7.5, "edad": 20, "traslado_ugb": 25, "uso_telefono": 4},
    "estudiante_9":  {"nombre": "Alejandro Gómez",     "materia": "Programación", "nota": 6.0, "edad": 19, "traslado_ugb": 50, "uso_telefono": 4},
    "estudiante_10": {"nombre": "Stiven Funes",     "materia": "Programación", "nota": 8.5, "edad": 19, "traslado_ugb": 35, "uso_telefono": 5},
    "estudiante_11": {"nombre": "STEVEN CASTILLO",  "materia": "Programación", "nota": 9.5, "edad": 22, "traslado_ugb": 10, "uso_telefono": 13},
    "estudiante_12": {"nombre": "ERNESTO CORDOVA ",     "materia": "Programación", "nota": 7.0, "edad": 20, "traslado_ugb": 55, "uso_telefono": 4},
    "estudiante_13": {"nombre": "Daniel Parada",    "materia": "Programación", "nota": 8.0, "edad": 25, "traslado_ugb": 70, "uso_telefono": 6},
    "estudiante_14": {"nombre": "Odalis Valle ",    "materia": "Programación", "nota": 4.0, "edad": 19, "traslado_ugb": 80, "uso_telefono": 8},
    "estudiante_15": {"nombre": "Jaime Chafit",   "materia": "Programación", "nota": 7.5, "edad": 19, "traslado_ugb": 30, "uso_telefono": 6},
    "estudiante_16": {"nombre": "Maricela Renderos ",    "materia": "Programación", "nota": 9.0, "edad": 21, "traslado_ugb": 100, "uso_telefono": 4},
    "estudiante_17": {"nombre": "Roberto Diaz",   "materia": "Programación", "nota": 6.5, "edad": 21, "traslado_ugb": 45, "uso_telefono": 6},
    "estudiante_18": {"nombre": "Misael Villafranco ",   "materia": "Programación", "nota": 8.0, "edad": 18, "traslado_ugb": 25, "uso_telefono": 9},
    "estudiante_19": {"nombre": "Jose Garcia", "materia": "Programación", "nota": 5.0, "edad": 22, "traslado_ugb": 65, "uso_telefono": 5},
    "estudiante_20": {"nombre": "Herber Marin",   "materia": "Programación", "nota": 8.5, "edad": 20, "traslado_ugb": 40, "uso_telefono": 5}
}

guardar_dataset_json(dataset_ugb)
dataset_cargado = cargar_dataset_json()

sistema = SistemaEstadistico(dataset_cargado)

sistema.calcular_todo("nota")
sistema.calcular_todo("traslado_ugb")
sistema.calcular_todo("uso_telefono")
sistema.calcular_todo("edad")

explicar_diferencias()
```

### **3.	Sistema estadístico, calcule: (20%)**

```python
from functools import reduce

class estadistica:
  def media(self, serie):
    suma = reduce(lambda acum, num: acum+ float(num), serie, 0)
    return suma/len(serie)

  def varianza(self, serie):
    media = self.media(serie)
    #4 + 0 + 4 + 0 + 0 + 16= 24/6 = 4.0
    suma = reduce(lambda acum, num: acum + (float(num)-media)**2, serie, 0)
    return suma/len(serie)

  def desviacion(self, serie):
    return self.varianza(serie)**0.5

  def media_armonica(self, serie):
    suma = reduce(lambda acum, num: acum + 1/float(num), serie, 0)
    return len(serie)/suma

  def fercuencia(self, serie):
    frecuencia = {}
    for num in serie:
      num_float = float(num)
      if num_float in frecuencia:
        frecuencia[num_float] += 1
      else:
        frecuencia[num_float] = 1
    return frecuencia

  def media_aritmetica(self, serie):
    suma = reduce(lambda acum, num: acum + float(num), serie, 0)
    return suma / len(serie)

  def moda(self, serie):
    frecuencia_datos = self.fercuencia(serie)
    max_frecuencia = max(frecuencia_datos.values())
    modas = [num for num, freq in frecuencia_datos.items() if freq == max_frecuencia]
    return modas

  def desviacion_tipica(self, serie):
    import math
    valor_varianza = self.varianza(serie)
    return math.sqrt(valor_varianza)

  def rango(self, serie):
    numeros = [float(num) for num in serie]
    return max(numeros) - min(numeros)

    objEstadistica = estadistica()
serie = input("Serie: ").split(",")
print("Media: ", objEstadistica.media(serie))
print("Varianza: ", objEstadistica.varianza(serie))
print("Desviacion: ", objEstadistica.desviacion(serie))
print("Media armonica: ", objEstadistica.media_armonica(serie))
print("Frecuencia: ", objEstadistica.fercuencia(serie))
print("Media aritmetica: ", objEstadistica.media_aritmetica(serie))
print("Moda: ", objEstadistica.moda(serie))
print("Desviacion tipica: ", objEstadistica.desviacion_tipica(serie))
print("Rango: ", objEstadistica.rango(serie))
```

# 2.	Desarrolle el conversor de Área (Superficie) (El Salavador) haciendo uso de los diccionarios y las funciones de usuario en Python que contenga las siguientes opciones: (30%)

```python
"""
Conversor de Área (Superficie) - El Salvador
Unidades: Pie², Vara², Yarda², Metro², Tareas, Manzana, Hectárea
"""

# Diccionario con los factores de conversión de cada unidad a metros cuadrados (m²)
# Estos son los valores de referencia usados comúnmente en El Salvador
FACTORES_A_M2 = {
    1: 0.092903,      # Pie cuadrado
    2: 0.698737,      # Vara cuadrada
    3: 0.836127,      # Yarda cuadrada
    4: 1.0,           # Metro cuadrado (unidad base)
    5: 582.4133,      # Tarea (1 manzana = 12 tareas)
    6: 6988.96,       # Manzana (10 000 varas²)
    7: 10000.0        # Hectárea
}

# Diccionario con los nombres visibles de cada unidad, para el menú y resultados
NOMBRES_UNIDADES = {
    1: "Pie Cuadrado",
    2: "Vara Cuadrada",
    3: "Yarda Cuadrada",
    4: "Metro Cuadrado",
    5: "Tareas",
    6: "Manzana",
    7: "Hectárea"
}


def mostrar_menu():
    """Muestra el menú numerado de unidades de área."""
    print("\n===== CONVERSOR DE ÁREA (SUPERFICIE) - EL SALVADOR =====")
    for numero, nombre in NOMBRES_UNIDADES.items():
        print(f"{numero}. {nombre}")
    print("0. Salir")


def pedir_unidad(mensaje):
    """Pide el número de una unidad y valida que exista en el diccionario."""
    while True:
        try:
            opcion = int(input(mensaje))
            if opcion in NOMBRES_UNIDADES:
                return opcion
            print("Opción inválida. Ingrese un número entre 1 y 7.")
        except ValueError:
            print("Debe ingresar un número.")


def pedir_valor_numerico(mensaje):
    """Pide un número y valida que sea un valor numérico positivo."""
    while True:
        try:
            valor = float(input(mensaje))
            if valor < 0:
                print("El valor no puede ser negativo.")
                continue
            return valor
        except ValueError:
            print("Debe ingresar un número válido.")


def convertir_area(valor, unidad_origen, unidad_destino):
    """
    Convierte un valor numérico desde una unidad de área hacia otra,
    usando los metros cuadrados como unidad intermedia (base).
    """
    valor_en_m2 = valor * FACTORES_A_M2[unidad_origen]
    valor_convertido = valor_en_m2 / FACTORES_A_M2[unidad_destino]
    return valor_convertido


def main():
    """Función principal: controla el ciclo del programa."""
    while True:
        mostrar_menu()

        unidad_origen = pedir_unidad("\nElija la unidad de origen (0 para salir): ")
        if unidad_origen == 0:
            print("Saliendo del conversor. ¡Hasta luego!")
            break

        valor = pedir_valor_numerico(f"Ingrese el valor en {NOMBRES_UNIDADES[unidad_origen]}: ")

        unidad_destino = pedir_unidad("Elija la unidad de destino (1-7): ")
        if unidad_destino == 0:
            print("Saliendo del conversor. ¡Hasta luego!")
            break

        resultado = convertir_area(valor, unidad_origen, unidad_destino)

        print(f"\n{valor} {NOMBRES_UNIDADES[unidad_origen]} equivalen a "
              f"{resultado:.4f} {NOMBRES_UNIDADES[unidad_destino]}")

        continuar = input("\n¿Desea realizar otra conversión? (s/n): ").strip().lower()
        if continuar != "s":
            print("Saliendo del conversor. ¡Hasta luego!")
            break


if __name__ == "__main__":
    main()
```

```python
"""
Calculadora del IMPUESTO A LAS ACTIVIDADES ECONÓMICAS
El usuario introduce el monto de su actividad económica y el programa
calcula el impuesto a pagar según la tabla de tramos vigente.
"""

# Diccionario con la tabla de tramos.
# Cada tramo guarda: (desde, hasta, precio_fijo, adicional_por_millar)
TABLA_IMPUESTO = {
    1:  (0.01,        500.0,       1.5,   0.00),
    2:  (500.01,      1000.0,      1.5,   3.00),
    3:  (1000.01,     2000.0,      3.0,   3.00),
    4:  (2000.01,     3000.0,      6.0,   3.00),
    5:  (3000.01,     6000.0,      9.0,   2.00),
    6:  (8000.01,     18000.0,     15.0,  2.00),
    7:  (18000.01,    30000.0,     39.0,  2.00),
    8:  (30000.01,    60000.0,     63.0,  1.00),
    9:  (60000.01,    100000.0,    93.0,  0.80),
    10: (100000.01,   200000.0,    125.0, 0.70),
    11: (200000.01,   300000.0,    195.0, 0.60),
    12: (300000.01,   400000.0,    255.0, 0.45),
    13: (400000.01,   500000.0,    300.0, 0.40),
    14: (500000.01,   1000000.0,   340.0, 0.30),
    15: (1000000.01,  99999999.0,  490.0, 0.18),
}


def buscar_tramo(monto):
    """
    Busca en la tabla el tramo al que pertenece el monto ingresado.
    Devuelve la tupla (desde, hasta, precio, adicional) o None si el
    monto no cae dentro de ningún tramo definido.
    """
    for datos in TABLA_IMPUESTO.values():
        desde, hasta, precio, adicional = datos
        if desde <= monto <= hasta:
            return datos
    return None


def calcular_impuesto(monto, tramo):
    """
    Calcula el impuesto a pagar aplicando la fórmula:
    ((monto - desde) / 1000) * adicional + precio
    """
    desde, hasta, precio, adicional = tramo
    impuesto = ((monto - desde) / 1000) * adicional + precio
    return round(impuesto, 2)


def pedir_monto():
    """Pide al usuario el monto de la actividad económica y valida que sea válido."""
    while True:
        try:
            monto = float(input("Ingrese el monto de su actividad económica: $"))
            if monto <= 0:
                print("El monto debe ser mayor a cero. Intente de nuevo.\n")
                continue
            return monto
        except ValueError:
            print("Eso no parece un número. Intente de nuevo.\n")


def main():
    """Función principal: saluda, calcula el impuesto y permite repetir."""
    print("¡Hola! Vamos a calcular cuánto tenés que pagar de impuesto")
    print("según el monto de tu actividad económica.\n")

    while True:
        monto = pedir_monto()
        tramo = buscar_tramo(monto)

        if tramo is None:
            print("\nUps, ese monto no encaja en ningún tramo registrado "
                  "en la tabla. Verificá el valor e intentá de nuevo.\n")
        else:
            impuesto = calcular_impuesto(monto, tramo)
            print(f"\nPara una actividad económica de ${monto:,.2f}, "
                  f"el impuesto a pagar es de ${impuesto:,.2f}\n")

        continuar = input("¿Querés calcular otro monto? (s/n): ").strip().lower()
        print()
        if continuar != "s":
            print("Gracias por usar la calculadora. ¡Hasta luego!")
            break


if __name__ == "__main__":
    main()
```
